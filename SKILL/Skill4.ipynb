```python

import os

# Create main project directory
project_name = "VisionPort"

os.makedirs(project_name, exist_ok=True)

# Create project folders
folders = [
    "data",
    "models",
    "src",
    "notebooks",
    "outputs",
    "tests",
    "configs"
]

for folder in folders:
    os.makedirs(os.path.join(project_name, folder), exist_ok=True)

print("VisionPort project structure created!")

# Display folders
for root, dirs, files_list in os.walk(project_name):
    level = root.replace(project_name, "").count(os.sep)
    print("  " * level + os.path.basename(root) + "/")
```

```python
import torch

print("PyTorch version:", torch.__version__)

print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("CUDA version:", torch.version.cuda)
    print("GPU:", torch.cuda.get_device_name(0))
    print("Number of GPUs:", torch.cuda.device_count())
else:
    print("Running on CPU")
```

```python
import torch

# Create a 2D tensor
tensor = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print("Tensor:")
print(tensor)

print("Shape:", tensor.shape)
print("Dimensions:", tensor.ndim)
print("Data type:", tensor.dtype)
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

tensor = tensor.to(device)

print("Tensor device:", tensor.device)
print(tensor)
```

```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Create two matrices on GPU
A = torch.randn(1000, 1000, device=device)
B = torch.randn(1000, 1000, device=device)

# Matrix multiplication
C = torch.matmul(A, B)

if device.type == "cuda":
    torch.cuda.synchronize()

print("Matrix multiplication completed!")
print("Matrix A shape:", A.shape)
print("Matrix B shape:", B.shape)
print("Result shape:", C.shape)
print("Computation device:", C.device)
```

```python
if torch.cuda.is_available():

    print("GPU Memory Allocated:",
          round(torch.cuda.memory_allocated(0) / 1024**2, 2), "MB")

    print("GPU Memory Reserved:",
          round(torch.cuda.memory_reserved(0) / 1024**2, 2), "MB")

    print("GPU Memory Total:",
          round(torch.cuda.get_device_properties(0).total_memory / 1024**2, 2), "MB")
else:
    print("CUDA GPU not available")
```
